## 1. データ準備
アップロードされた`archive.zip`ファイルを解凍し、中身を確認します。このステップは、分析に利用するデータを特定し、どのような形式でデータが格納されているかを理解するために重要です。

```python
import zipfile
import os

# 解凍先のディレクトリを作成
extract_dir = './data'
os.makedirs(extract_dir, exist_ok=True)

# archive.zipを解凍
with zipfile.ZipFile('archive.zip', 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# 解凍されたファイルの一覧を表示
print(f"'{extract_dir}'に解凍されたファイル一覧:")
for root, dirs, files in os.walk(extract_dir):
    for file in files:
        print(os.path.join(root, file))
```

## 2. データ読み込みと概要確認

解凍されたCSVファイルをPandasのDataFrameとして読み込み、各データの先頭行と基本情報を確認します。このステップでは、データがどのような内容を含んでいるか、欠損値の有無、データ型などを把握し、後のデータクレンジングや分析計画の基礎を築きます。

**なぜこのステップを行うのか？**

*   **データ構造の理解**: 各ファイルの列名、レコード数、およびデータの種類（数値、テキスト、日付など）を把握します。
*   **データ品質の初期確認**: 欠損値の存在や、データ型の不一致など、初期段階で発見できるデータ品質の問題を特定します。
*   **分析方向性の検討**: データの概要から、どのような分析が可能か、どのような情報が得られそうかという仮説を立てるためのインスピレーションを得ます。

```python
import pandas as pd

# customer_details.csv を読み込む
customer_df = pd.read_csv('./data/customer_details.csv')
print("### customer_details.csv の概要 ###")
print("\n--- 先頭5行 ---")
print(customer_df.head())
print("\n--- 情報 ---")
print(customer_df.info())

print("\n" + "="*50 + "\n")

# basket_details.csv を読み込む
basket_df = pd.read_csv('./data/basket_details.csv')
print("### basket_details.csv の概要 ###")
print("\n--- 先頭5行 ---")
print(basket_df.head())
print("\n--- 情報 ---")
print(basket_df.info())
```

## 3. データクレンジングと前処理

ここでは、データ分析に適した形式にするために、データ型の変換とデータフレームの結合を行います。

**なぜこのステップを行うのか？**

*   **データ型の最適化**: `basket_date`は日付情報として扱う必要があるため、`datetime`型に変換します。これにより、時間軸での分析（例: 月ごとの売上トレンド）が可能になります。`customer_age`は年齢であり、通常は整数で表現されるため、`float`から`int`に変換することでメモリ効率の向上と直感的な理解を促します。
*   **データ統合**: `customer_df`と`basket_df`は`customer_id`を共通のキーとして持っています。これらを結合することで、各顧客の属性（性別、年齢、tenure）と購入履歴（購入日、購入商品、購入数）を紐付け、より多角的な分析を可能にします。

**「現場の泥臭いアクション」**

実務では、データ型が間違っていたり、結合キーが不完全だったりすることがよくあります。`basket_date`が`object`型であることは、日付として扱えないため、`pd.to_datetime`で正確に変換する必要があります。また、結合する際には、`customer_id`が両方のデータフレームに存在し、一意であることを確認することが重要です。

```python
# データの型変換
# basket_df の 'basket_date' を datetime 型に変換
basket_df['basket_date'] = pd.to_datetime(basket_df['basket_date'])
print("--- basket_df の 'basket_date' 型変換後 ---")
print(basket_df.info())

# customer_df の 'customer_age' を int 型に変換
# 欠損値がないことを確認済みのため、そのまま変換
customer_df['customer_age'] = customer_df['customer_age'].astype(int)
print("\n--- customer_df の 'customer_age' 型変換後 ---")
print(customer_df.info())

print("\n" + "="*50 + "\n")

# データフレームの結合
# なぜこのキー（ID）で結合したのか？
# customer_id は顧客を一意に識別するIDであり、両方のデータフレームに存在するため、
# 顧客の属性情報と購入履歴を紐付けるための最適なキーです。

# 結合によって消えてしまったデータ（欠落）はないか？
# 今回は両方のデータフレームに存在するcustomer_idのみを結合する内部結合(inner join)を選択します。
# これにより、購入履歴がない顧客や、顧客情報がない購入履歴は分析から除外されますが、
# 今回の目的（収益向上）においては、購買行動を伴うデータに焦点を当てるため、この選択は妥当です。
# 結合前にそれぞれのユニークなcustomer_idの数を数えて、結合後のデータ数を予測することも可能です。

# 結合前のユニークなcustomer_idの数
num_unique_customer_ids_customer = customer_df['customer_id'].nunique()
num_unique_customer_ids_basket = basket_df['customer_id'].nunique()

print(f"結合前の customer_df のユニークな customer_id 数: {num_unique_customer_ids_customer}")
print(f"結合前の basket_df のユニークな customer_id 数: {num_unique_customer_ids_basket}")

# 内部結合
merged_df = pd.merge(basket_df, customer_df, on='customer_id', how='inner')

print("\n### 結合後のデータフレームの概要 ###")
print("\n--- 先頭5行 ---")
print(merged_df.head())
print("\n--- 情報 ---")
print(merged_df.info())

# 結合後の行数を確認し、データ欠落の有無を推測
print(f"\n結合後の行数: {len(merged_df)}")
print(f"basket_dfの行数: {len(basket_df)}")
# merged_df の行数が basket_df の行数と同じか少ない場合、結合は適切に行われている可能性が高い。
# もしcustomer_dfのcustomer_idに重複がある場合、行数が増える可能性があるが、info()で確認した限り重複はなさそう。
```

## 4. データ統合の課題と対応（「泥臭いアクション」の具体例）

前処理のステップでデータフレームの結合を試みましたが、`merged_df`の行数が`basket_df`の15000行からわずか72行に激減していることが確認されました。これは、`customer_id`をキーとした結合において、両方のデータフレームに共通して存在する顧客IDが極めて少ないことを示唆しています。

**なぜこの問題が発生したのか？**

*   **データセットの不整合**: `customer_df`と`basket_df`は、それぞれ顧客情報と購入履歴を提供していますが、`customer_id`の体系が異なるか、あるいは意図的に一致しないように設計された異なる期間やソースのデータである可能性があります。例えば、`customer_df`は全ての登録顧客情報で、`basket_df`は特定のキャンペーン期間中の購入履歴のみで、かつそのキャンペーン対象顧客の一部しか`customer_df`に存在しない、といったケースが考えられます。
*   **データ漏洩**: 結合キーとなる`customer_id`に何らかの変換ミスや誤りがあった可能性も排除できません（今回はデータ型を確認済みですが、例えば先頭ゼロの有無などで文字列として異なる扱いになる場合など）。

**「データ構造の整合性」を疑う視点と、結合によって消えてしまったデータ（欠落）への考察**

今回、`customer_id`という共通の列名を持つデータを使用しましたが、そのIDが実際に同じエンティティ（顧客）を指しているか、という「データ構造の整合性」を疑う必要があります。結果として、13,871件のユニークな`customer_id`を持つ`basket_df`のほとんどの購入履歴が、`customer_df`の顧客情報と紐付けられない状況です。

この状況では、顧客の属性（性別、年齢、tenure）と具体的な購入行動を紐付けた広範な分析は困難です。実務では、このような場合、データ提供元に問い合わせてIDの整合性を確認したり、別の結合キーを探したり、あるいはデータを諦めて別々の分析を行うといった判断が求められます。

**今回の分析方針**

与えられたデータで収益向上を目指す分析を行うため、ここでは以下の戦略をとります。

1.  **`basket_df`に焦点を当てた販売・商品分析**: `customer_id`の結合がうまくいかないため、`basket_df`のみを用いて、全体的な販売トレンド、人気商品、購入頻度など、**直接的に収益に影響する販売行動**に関する分析を行います。これにより、「売上を上げるための分析」の一端を担います。
2.  **`customer_df`は顧客属性の全体像把握に留める**: `customer_df`は、性別、年齢、tenureといった顧客のデモグラフィック情報を独立して分析する際に活用します。個々のバスケットデータとの直接的な関連は限定的であることを前提とします。

まずは、`basket_df`から「収益を上げる」ための洞察を得るために、時系列分析や商品分析に必要な情報を抽出します。

```python
# customer_df と basket_df に共通して存在する customer_id の数を再確認
common_customer_ids = pd.Series(list(set(customer_df['customer_id']).intersection(set(basket_df['customer_id']))))

print(f"customer_df と basket_df に共通して存在するユニークな customer_id の数: {len(common_customer_ids)}")
print(f"これは basket_df のユニーク customer_id 数 ({basket_df['customer_id'].nunique()}) と比較して非常に少ないです。")

# merged_df の行数も72と非常に少ないため、このデータセット全体で顧客属性と購入履歴を紐づけるのは困難と判断します。
# そこで、basket_df を使って販売トレンドの分析を進めます。

# basket_date から年、月、曜日などの特徴量を抽出
basket_df['year'] = basket_df['basket_date'].dt.year
basket_df['month'] = basket_df['basket_date'].dt.month
basket_df['day_of_week'] = basket_df['basket_date'].dt.day_name()

print("\n--- 特徴量抽出後の basket_df 先頭5行 ---")
print(basket_df.head())
print("\n--- 特徴量抽出後の basket_df 情報 ---")
print(basket_df.info())
```

## 5. 販売トレンドと商品パフォーマンス分析

ここでは、`basket_df`を活用して、販売データから収益向上につながる洞察を得るための分析を行います。具体的には、購買の時系列トレンド、人気商品の特定、曜日ごとの販売傾向などを可視化し、潜在的な課題や機会を見つけ出します。

**なぜこの分析を行うのか？**

*   **収益の最大化**: 全体的な販売動向や人気商品を把握することで、在庫管理の最適化、プロモーション戦略の立案、新商品開発の方向性などを検討し、売上と利益の最大化を目指します。
*   **機会費用の最小化**: 売れない商品の在庫リスクや、販売機会の損失を防ぐための情報を提供します。
*   **具体的な施策への落とし込み**: 分析結果から、「いつ、どの商品を、どのように売るか」という具体的なアクションプランを導き出すための根拠とします。

**分析項目:**

1.  月ごとの販売トレンド
2.  曜日ごとの販売傾向
3.  売上上位商品（`product_id`別）

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# 1. 月ごとの販売トレンド
# basket_count の合計を月ごとに集計
monthly_sales = basket_df.groupby(['year', 'month'])['basket_count'].sum().reset_index()
monthly_sales['year_month'] = monthly_sales['year'].astype(str) + '-' + monthly_sales['month'].astype(str).str.zfill(2)
monthly_sales = monthly_sales.sort_values('year_month')

plt.figure(figsize=(12, 6))
sns.lineplot(data=monthly_sales, x='year_month', y='basket_count', marker='o')
plt.title('月ごとの合計販売数トレンド')
plt.xlabel('年月')
plt.ylabel('合計販売数')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("このグラフは、時間経過に伴う月ごとの販売数の推移を示しています。これにより、季節性、トレンド、異常値などを把握し、将来の販売予測やマーケティングキャンペーンの計画に役立てることができます。例えば、特定の月に販売数が急増していれば、その時期のプロモーション強化を検討できます。")

# 2. 曜日ごとの販売傾向
day_of_week_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
day_sales = basket_df.groupby('day_of_week')['basket_count'].sum().reindex(day_of_week_order).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=day_sales, x='day_of_week', y='basket_count', palette='viridis')
plt.title('曜日ごとの合計販売数')
plt.xlabel('曜日')
plt.ylabel('合計販売数')
plt.tight_layout()
plt.show()

print("このグラフは、曜日ごとの合計販売数を示しています。どの曜日に販売が活発であるかを把握することで、人員配置の最適化、特定の曜日におけるプロモーションの強化、配送スケジュールの調整など、オペレーション効率化と売上向上に繋がる施策を検討できます。")

# 3. 売上上位商品（basket_countの合計）
top_products = basket_df.groupby('product_id')['basket_count'].sum().nlargest(10).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=top_products, x='basket_count', y='product_id', orient='h', palette='magma')
plt.title('販売数上位10商品')
plt.xlabel('合計販売数')
plt.ylabel('商品ID')
plt.tight_layout()
plt.show()

print("このグラフは、最も販売数の多い上位10商品を示しています。これらの商品は顧客からの需要が高いと推測できるため、これらの商品の在庫を十分に確保すること、関連商品をレコメンドすること、またはプロモーションの中心に据えることで、収益のさらなる増加を期待できます。")
```

### 課題定義

## 7. 提案と投資対効果 (ROI)

特定した課題に対して、具体的な解決策と、その実施による期待効果（投資対効果）を提案します。

### 提案1: 季節性・曜日性に合わせたプロモーションと在庫管理の最適化

**課題**: 月や曜日によって販売数に変動があるが、販売機会を十分に活用できていない。

**仮説**: 販売が低調な時期にターゲットを絞ったプロモーションを実施し、販売ピーク時には在庫を適切に確保することで、全体の販売数を増加させることができる。

**具体的な施策（現場の泥臭いアクション）**:

## 8. 顧客行動からの利益向上分析（RFM分析）

これまでの分析は、主に商品や時間軸での販売トレンドに焦点を当てていました。しかし、利益を向上させるためには、顧客一人ひとりの行動を理解し、それに合わせた戦略を立てることが非常に重要です。

ここでは、RFM（Recency, Frequency, Monetary）分析という手法を用いて、顧客を購買行動の観点からセグメント化し、異なる着眼点から利益向上策を探ります。

**RFM分析とは？**

*   **Recency（最終購入日からの日数）**: 最近商品を購入した顧客ほど、再び購入する可能性が高いと考えられます。
*   **Frequency（購入頻度）**: 頻繁に商品を購入する顧客は、ブランドロイヤリティが高い可能性があり、今後も購入を継続する可能性が高いです。
*   **Monetary（購入金額）**: 一度に多額の商品を購入する顧客は、LTV（顧客生涯価値）が高い傾向にあります。今回は`basket_count`を金額の代理指標として利用します。

**なぜこの分析を行うのか？**

*   **顧客セグメンテーション**: 顧客を数種類のグループに分類することで、各グループの特性を理解し、それぞれに最適化されたマーケティング戦略を立案できます。
*   **パーソナライズされたアプローチ**: 例えば、「最近購入していないが、過去に頻繁に購入していた顧客」と「最近購入したが、購入金額が低い顧客」では、取るべきアプローチが異なります。RFM分析により、このようなターゲット顧客を明確にします。
*   **ROIの最大化**: 無差別にプロモーションを行うよりも、顧客の行動パターンに基づいてターゲットを絞ることで、マーケティング費用対効果（ROI）を高めることができます。

```python
import datetime as dt

# 最新のバスケット日付を基準日とする
# basket_df['basket_date']はすでにdatetime型に変換済み
latest_date = basket_df['basket_date'].max() + dt.timedelta(days=1) # 基準日は最新購入日の翌日とする

# RFMの計算
rfm_df = basket_df.groupby('customer_id').agg(
    Recency=('basket_date', lambda date: (latest_date - date.max()).days), # 最新購入日からの日数
    Frequency=('basket_date', 'count'), # 購入回数 (ここではバスケットの行数を使用)
    Monetary=('basket_count', 'sum') # 購入点数の合計を金額の代理とする
).reset_index()

print("### RFM分析結果の先頭5行 ###")
print(rfm_df.head())

print("\n--- RFM指標の統計量 ---")
print(rfm_df[['Recency', 'Frequency', 'Monetary']].describe())

# RFM指標の分布を可視化
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.histplot(rfm_df['Recency'], bins=30, kde=True)
plt.title('Recencyの分布')
plt.xlabel('最終購入日からの日数')
plt.ylabel('顧客数')

plt.subplot(1, 3, 2)
sns.histplot(rfm_df['Frequency'], bins=30, kde=True)
plt.title('Frequencyの分布')
plt.xlabel('購入回数')
plt.ylabel('顧客数')

plt.subplot(1, 3, 3)
sns.histplot(rfm_df['Monetary'], bins=30, kde=True)
plt.title('Monetaryの分布')
plt.xlabel('合計購入点数')
plt.ylabel('顧客数')

plt.tight_layout()
plt.show()

print("これらのRFM指標の分布グラフは、顧客の購買行動パターンを全体的に把握するのに役立ちます。例えば、Recencyの分布から多くの顧客が最近購入しているか、離反しつつあるかがわかります。FrequencyとMonetaryの分布からは、購入頻度が高い顧客や高額購入者がどの程度いるかを把握し、セグメンテーションの基準を検討する材料となります。")
```

*   **アクション**: 購入点数に応じた割引制度、テーマ別商品セットの提案、「あと〇〇円で送料無料」の表示。
*   **期待効果**: 平均購入点数増加、顧客あたりの平均購入金額増加、直接的な売上向上。

## 今後の展望

*   **データ不整合の根本原因解明**: `customer_id`の不整合について、データ提供元との連携を通じて原因を特定し、データ統合の改善を図ります。
*   **ROIの具体的な数値化**: 各提案の期待効果とコストについて、実際のビジネスデータに基づいて具体的な数値を算出し、より精度の高いROI分析を行います。
*   **商品マスターデータの統合**: `product_id`に対応する商品名やカテゴリ情報を取得し、より詳細な商品分析（例：カテゴリごとの売上貢献度、商品間の関連性分析）を可能にします。
